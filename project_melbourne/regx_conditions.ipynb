```python
import pandas as pd
import re
```

```python
df = pd.read_csv('../data_base/house.csv')
```

```python
filtro1 = df.loc[(df['Rooms'] == 3) & (df['Type'] == 'h') & (df['Price'] <= 400000)] # Filtra  através de condicionais
```

```python
filtro2 = df.loc[df['Address'].str.contains('Turner St | Turner Rd', flags=re.I)] # Filtra por conjunto de caracteres contidos na coluna 
# - 'Turner St | Turner Rd' -> Turner St or Turner Rd 
# - flags=re.I -> Para ignorar letras mainúsculas e minúscúlas
```

```python
filtro3 = df.loc[df['Address'].str.contains('^59', flags=re.I) | (df['Price'] <= 500000)] # Filtra por conjunto de caracteres contidos na coluna - A casa será com o número 59 ou com o valor maior ou igual á 500000
# - '^59' -> ^ indica que verificará o primeiro carácter
# - flags=re.I -> Para ignorar letras mainúsculas e minúscúlas

filtro3.iloc[0:10] # Mostrando do index 0 até o 4 
```
