```python
import numpy as np
import sys
```

```python
a = [1,2,3,4,5] # Lista convencional
b = np.array([1,2,3,4,5]) # Numpy
```

```python
print(a) # Lista convencional, com virgula
print(b) # Observe que não há virgulas
```

```python
sys.getsizeof(a) # Retorna o quanto 'a' ocupa em bytes
```

```python
b.nbytes #  Retorna o quanto 'b' ocupa em bytes
```
