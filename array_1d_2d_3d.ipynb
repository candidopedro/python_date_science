```python
import numpy as np
```

```python
# Array 1D
d1 = np.array([1,5,4])
print(f'{d1.shape}\n{d1.ndim}\n{d1.dtype}\n')
print(d1)
```

```python
# Array 2D
d2 = np.array([[1,5,4],[4,3,7]])
print(f'{d2.shape}\n{d2.ndim}\n{d2.dtype}\n')
print(d2)
```

```python
# Array 3D
d3 = np.array([[1,5,4],[4,3,7],[6,2,8]])
print(f'{d3.shape}\n{d3.ndim}\n{d3.dtype}\n')
print(d3)

```
