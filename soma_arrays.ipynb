```python
import numpy as np
```

```python
a = [1,7,4,6] 
b = [4,6,3,2]
```

```python
result = a + b # Somando listas convencionais, juntando as duas
result
```

```python
c = np.array([1,7,4,6]) 
d = np.array([4,6,3,2])
result_array = np.add(c, d) # Faz a soma dos itens pelo index: index 1 -> '1+4 = 5' index 2 -> '7+6 = 13' index 3 -> '4+3 = 7' index 4 -> '6+2 = 8'
print(result_array)
```
