```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
```

```python
gas = pd.read_csv('../data_base/gas_prices.csv')
```

```python
plt.title('Price is US$') # <- Adiciona um título ao gráfico

list_country = ['Canada', 'USA', 'UK', 'Japan'] # <- Lista de alguns paises existentes no CSV. 

for country in gas: # <- Irá percorrer o CSV através do for
    if country in list_country: # <- Se o pais estiver na lista será mostrado no gráfico
        plt.plot(gas['Year'], gas[country], marker='.', label = country) # marker='.' -> Coloca a uma linha pontilhada em cada elemento do gráfico

# Eixo X ->
plt.xticks(gas['Year'][::3]) # <- Ajusta o eixo X, mostrando de 2 em 2 anos
plt.xlabel('Years')

# Eixo Y ->
plt.ylabel('US$')

plt.legend() # <- Mostra a idendificação de cada linha de acordo com a COR
plt.show() # <- Mostra a tabela
```
