```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
```

```python
gas = pd.read_csv('../data_base/gas_prices.csv')
```

```python
plt.title('Price is US$') # <- Adiciona um título ao gráfico

plt.plot(gas['Year'], gas['Canada'], 'r.--', label = 'Canada') # <- Retorna um gráfico com TMEPO(Year/Anos) em função do VALOR do combustível no Canada
plt.plot(gas['Year'], gas['USA'], 'b.--', label = 'USA')
plt.plot(gas['Year'], gas['UK'], 'k.--', label = 'UK')

# Eixo X ->
plt.xticks(gas['Year'][::3]) # <- Ajusta o eixo X, mostrando de 2 em 2 anos
plt.xlabel('Years')

# Eixo Y ->
plt.ylabel('US$')

plt.legend() # <- Mostra a idendificação de cada linha de acordo com a COR
plt.show() # <- Mostra a tabela
```
