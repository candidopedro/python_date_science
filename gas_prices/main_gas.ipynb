```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
```

```python
gas = pd.read_csv('../data_base/gas_prices.csv')
```

```python
plt.title('Valor do combustível em US$') # <- Adiciona um título ao gráfico

plt.plot(gas['Year'], gas['Canada'], label = 'Canada') # <- Retorna um gráfico com TMEPO(Year/Anos) em função do VALOR do combustível no Canada
plt.plot(gas['Year'], gas['USA'], label = 'USA')
plt.plot(gas['Year'], gas['UK'], label = 'UK')

plt.xticks(gas['Year'][::2]) # <- Ajusta o eixo X, mostrando de 2 em 2 anos

plt.legend() # <- Mostra a idendificação de cada linha de acordo com a COR
plt.show() # <- Mostra a tabela
```
